```python
import numpy as np
import pandas as pd

numbers = np.array([8, 12, 18, 24, 30])
print("Array:", numbers)
print("Array + 5:", numbers + 5)
print("Array × 2:", numbers * 2)
print("First three values:", numbers[:3])
print("Values above 15:", numbers[numbers > 15])

data = {
    "Name": ["Neha", "Karan", "Priya", "Vikram"],
    "Age": [22, 24, 21, 23],
    "Score": [91, 79, 87, 94]
}

df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)

print("\nScore column:")
print(df["Score"])

print("\nFirst two rows:")
print(df.iloc[:2])

print("\nNames and scores for the first three rows:")
print(df.loc[:2, ["Name", "Score"]])

print("\nStudents with scores above 80:")
print(df[df["Score"] > 80])
```
