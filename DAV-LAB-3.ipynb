```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "StudyHours": [2, 4, 6, 3, 7, 8, 5, 9, 1, 6],
    "Marks": [52, 68, 81, 61, 85, 92, 76, 96, 45, 73],
    "Attendance": [62, 74, 88, 69, 91, 96, 82, 98, 54, 79],
    "SleepHours": [6, 7, 8, 6, 7, 8, 7, 8, 5, 7]
})

print("First five rows:")
display(df.head())

print("Dataset size:", df.shape)

print("Column names:")
print(df.columns.tolist())

print("Data types:")
print(df.dtypes)

print("Descriptive statistics:")
display(df.describe())

sns.histplot(data=df, x="Marks", bins=5, kde=True)
plt.title("Distribution of College Marks")
plt.show()

sns.boxplot(data=df, x="Marks")
plt.title("Marks Box Plot")
plt.show()

sns.scatterplot(data=df, x="StudyHours", y="Marks", hue="Attendance")
plt.title("Study Hours vs Marks")
plt.show()

sns.pairplot(df)
plt.show()

sns.heatmap(df.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
