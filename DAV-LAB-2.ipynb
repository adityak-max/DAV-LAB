```python
import pandas as pd

df = pd.DataFrame({
    "Student_Name": ["Rahul", "Sneha", "Arjun", "Sneha", "Priya", "Karan", "Rahul", "Meera"],
    "Age": [19, 20, None, 20, 21, 19, 19, None],
    "City": ["Mysore", "Bangalore", "Hubli", "Bangalore", "Mangalore", "Mysore", "Mysore", "Hubli"],
    "Branch": ["CSE", "ISE", "ECE", "ISE", "CSE", "ECE", "CSE", "ISE"],
    "Marks": [85, 78, 92, 78, None, 74, 85, 88],
    "Credits_Earned": [24, 22, 26, 22, 20, None, 24, 25]
})

print("First five rows:")
display(df.head())

print("Dataset size:", df.shape)

print("Column names:")
print(df.columns.tolist())

print("Data types:")
print(df.dtypes)

print("Descriptive statistics:")
display(df.describe(include="all"))

print("Missing values in each column:")
print(df.isnull().sum())

print("Total missing values:")
print(df.isnull().sum().sum())

print("Number of duplicate rows:")
print(df.duplicated().sum())

clean_df = df.drop_duplicates()

print("Dataset after removing duplicates:")
display(clean_df)

print("Missing values after removing duplicates:")
print(clean_df.isnull().sum())
```
